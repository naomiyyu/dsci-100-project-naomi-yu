#### Task 4.1: Calculate summary statistics of numeric variables (2 points)

- Use any functions from the `tidyverse` to create **one** table that outputs the mean, minimum, and maximum of **all** numeric columns in your data, dropping the missing values if they exist.

```r
library(tidyverse)
basketball_summary <- basketball_tidy |> 
    summarize(total_losses_min = min(total_losses, na.rm = TRUE), total_losses_max = max(total_losses, na.rm = TRUE), total_losses_mean = mean(total_losses, na.rm = TRUE), total_wins_min = min(total_wins, na.rm = TRUE), total_wins_max = max(total_wins, na.rm = TRUE), total_wins_mean = mean(total_wins, na.rm = TRUE))
basketball_summary
```

#### Task 4.2: Determine the amount of missingness (2 points)

- Use your `dplyr` knowledge to create **one** table that calculates the number of missing values for each column in the data set (both numeric and categorical).

```r
basketball_missingness <- basketball |>
    summarize(na_first_game_at_home = sum(is.na(first_game_at_home)), na_total_losses = sum(is.na(total_losses)), na_total_wins = sum(is.na(total_wins)))
basketball_missingness
```

#### Task 4.3: Summarize over levels of a categorical variable (2 points)

```r
basketball_summary <- basketball_tidy |> 
    group_by(first_game_at_home) |>
    summarize(total_losses_mean = mean(total_losses, na.rm = TRUE), total_wins_mean = mean(total_wins, na.rm = TRUE))
basketball_summary
```

#### Task 4.4: Plot the distribution of two relevant variables in the raw data (4 points)

```r
### EDIT THIS CODE CELL FOR PLOT 1 ### ONE SCATTER PLOT TO SEE POINTS RELATIONSHIP
basketball_plot_relationship_variables <- basketball_tidy |> 
    ggplot(aes(x = total_wins, y = total_losses)) +
    geom_point(alpha = 0.4) +
    labs(x = "Total Wins", y = "Total Losses", title = "Relationship of Total Losses and Total Wins of Each Team") +
    theme(text = element_text(size = 15))
basketball_plot_relationship_variables
```

```r
### EDIT THIS CODE CELL FOR PLOT 2 ### ONE BAR PLOT TO SEE HOW MUCH Y VALUE VS N VALUE FOR WHETHER GAME PLAYED AT HOME
basketball_plot_first_game_at_home_prop <- basketball_tidy |> 
    ggplot(aes(x = first_game_at_home, fill = first_game_at_home)) + 
    geom_bar(stat = "count") +
    labs(x = "Is the First Game Played At Home", y = "Number of Teams", fill = "Is the First Game Played At Home", title = "Proportion of Number of Teams that Played Their First Game At Home or Not") +
    theme(text = element_text(size = 15), legend.position = "bottom")
basketball_plot_first_game_at_home_prop
```

Visualization 1: This plots shows the relationship between the two predictor variables, total losses and total wins, and the distribution of the point regarding the response variable, is the first game played at home. The two predictors have a strong negative relationship, where total losses decrease, as total wins increase. Also, most data points on the left, lower in total wins and higher in total losses, has the response variable N (first game not at home), and most data points on the right, higher in total wins and lower in total losses, has the response variable Y (first game played home).

```r
### EDIT THIS CODE CELL FOR VISUALIZATION 1### ONE SCATTER PLOT WITH CATEGORYS AS COLOURS
basketball_plot_category <- basketball_tidy |> 
    ggplot(aes(x = total_wins, y = total_losses, color = first_game_at_home)) +
    geom_point(alpha = 0.3) +
    labs(x = "Total Wins", y = "Total Losses", color = "Is the First Game Played At Home", title = "Relationship of Total Losses and Total Wins of Each Team") +
    theme(text = element_text(size = 15), legend.position = "bottom")
basketball_plot_category
```

Visualization 2: This plot compares the average total wins and total losses for teams that played their first game at their home court, and the average total wins and total losses for teams that played their first game at a different teams court. This is useful to compare whether playing the first game at the home court changes the proportion of total wins and total losses in general. As we can see, playing the first game at home correlates to a higher total wins to losses ratio compared to not playing the first game at home.

```r
### EDIT THIS CODE CELL FOR VISUALIZATION 2### ONE BAR PLOT SHOWING PROPORTION OF MEAN TOTAL WINS AND LOSSES
library(tidyverse)
basketball_summary_longer <- basketball_summary |> 
    pivot_longer(cols = total_losses_mean:total_wins_mean, names_to = "win_loss", values_to = "mean")

basketball_summary_bar <- basketball_summary_longer |>
    ggplot(aes(x = win_loss, y = mean, fill = first_game_at_home)) +
    geom_bar(stat = "identity", position = "dodge") +
    labs(x = "Total Win or Total Loss", fill = "Is the First Game at Home", y = "Mean", title = "Proportion of Mean Total Wins/Losses for First Game Played at Home vs Not") + 
    theme(text = element_text(size = 15), legend.position = "bottom")

basketball_summary_bar
```
